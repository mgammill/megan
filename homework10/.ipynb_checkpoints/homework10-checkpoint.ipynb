```python
for name in df.columns:
    if name != "Planet" and name !="Ring System?" and name != "Global Magnetic Field?":
        df[name] = pd.to_numeric(df[name], errors="coerce")
```

```python
# 2.1.3.3 Answer the following:
# (a) What are the three different data types now in the DataFrame?
# The three different data types in the Data Frame are objects, floats (float64), and integers (int64).
# (b) Define each one (e.g. float64).
# An object is an instance of class, which contain data and perform actions. A float64 is a 
# decimal value, which is comprises of 64 bits (refering to the size of the data). An int64 is 
# an integer value (a whole number) also comprised of 64 bits of memory.
# (c) Which columns are still object, and why?
# The columns that are still object are Planet, Ring System?, and Global Magnetic Field?. 
# These columns still have the "object" datatype because they hold strings, rather than numerical
# values like floats or integers.

for name in df.columns:
    type = df[name].dtypes
    if type == object:
        print(name)
    
```

```python
def attach_units(original_col, new_col, unit):
    new_values = []
    for value in df[original_col]:
        value = value * unit
        new_values.append(value)
    df[original_col] = new_values
    df.rename(columns={original_col: new_col}, inplace=True)

attach_units("Mass (10^24kg)", "Mass (kg)", 1e24 * u.kg)
print(df["Mass (kg)"])
```

```python
# for name in df.columns:
#     print(name)

semi_major_axis = []
for peri, aphe in zip(df["Perihelion (10^6 km)"], df["Aphelion (10^6 km)"]):
    sma = (peri + aphe) / 2
    sma = sma * 1e6 * u.km
    semi_major_axis.append(sma)
```

```python
df.insert(11, "Semi-Major Axis (km)", semi_major_axis, allow_duplicates=False)

for name in df.columns:
    print(name)
```

```python
print(df["Semi-Major Axis (km)"])
```

```python
print(df["Orbital Period (days)"])

orbital_period_yrs = []
for value in df["Orbital Period (days)"]:
    value_d = value * u.d
    value_yr = value_d.to(u.yr)
    orbital_period_yrs.append(value_yr)
df["Orbital Period (days)"] = orbital_period_yrs
df.rename(columns={"Orbital Period (days)": "Orbital Period (years)"}, inplace=True)

for name in df.columns:
    print(name)

print(df["Orbital Period (years)"])
    
```

```python
# 2.3.3
# Define AU in your own words.
# AU is a unit used to describe very big distances in space. It measures the approximate 
# distance between Earth and the Sun, about 93 million miles. 
au = const.au
print(au)
au_km = au.to(u.km)
print(au_km)
# There are 149597870.7 km in one AU.
```

```python
def add_units(column_name, units):
    new_values = []
    for value in df[column_name]:
        value_units = value * units
        new_values.append(value_units)
    df[column_name] = new_values
    
def change_units(column_name, new_column_name, new_units):
    new_values = []
    for value in df[column_name]:
        new_value = value.to(new_units)
        new_values.append(new_value)
    df[column_name] = new_values
    df.rename(columns={column_name: new_column_name}, inplace=True)
```

```python
for value in df["Diameter (km)"]:
    print(value)

add_units("Diameter (km)", u.km)

for value in df["Diameter (km)"]:
    print(value)
```

```python
change_units("Diameter (km)", "Diameter (AU)", u.au)

for value in df["Diameter (AU)"]:
    print(value)
```

```python
# for value in df["Escape Velocity (km/s)"]:
#     print(value)

add_units("Escape Velocity (km/s)", u.km)
change_units("Escape Velocity (km/s)", "Escape Velocity (AU/s)", u.au)

new_values = []
for value in df["Escape Velocity (AU/s)"]:
    value_units = value /  u.s
    new_values.append(value_units)
df["Escape Velocity (AU/s)"] = new_values

# for value in df["Escape Velocity (AU/s)"]:
#     print(value)
```

```python
def add_1e6km(column_name):
    new_values = []
    for value in df[column_name]:
        value_units = value * 1e6 * u.km
        new_values.append(value_units)
    df[column_name] = new_values

add_1e6km("Distance from Sun (10^6 km)")

change_units("Distance from Sun (10^6 km)", "Distance from Sun (AU)", u.au)

for value in df["Distance from Sun (AU)"]:
    print(value)
```

```python
add_1e6km("Perihelion (10^6 km)")
change_units("Perihelion (10^6 km)", "Perihelion (AU)", u.au)

add_1e6km("Aphelion (10^6 km)")
change_units("Aphelion (10^6 km)", "Aphelion (AU)", u.au)    

for value in df["Perihelion (AU)"]:
    print(value)

for value in df["Aphelion (AU)"]:
    print(value)
```

```python
for value in df["Semi-Major Axis (km)"]:
    print(value)

change_units("Semi-Major Axis (km)", "Semi-Major Axis (AU)", u.au)

for value in df["Semi-Major Axis (AU)"]:
    print(value)
```

```python
for value in df["Orbital Velocity (km/s)"]:
    print(value)

add_units("Orbital Velocity (km/s)", u.km)
change_units("Orbital Velocity (km/s)", "Orbital Velocity (AU/s)", u.au)

new_values = []
for value in df["Orbital Velocity (AU/s)"]:
    value_units = value / u.s
    new_values.append(value_units)
    
df["Orbital Velocity (AU/s)"] = new_values

for value in df["Orbital Velocity (AU/s)"]:
    print(value)
```
