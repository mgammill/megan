```python
import pandas as pd
import numpy as np
from IPython.display import display
```

```python
df = pd.read_csv("../../course_assignments/homework8/state_data.csv")
print(df.shape) # There are 52 rows and 4 columns
```

```python
print(df.head())
print(df.tail())
# The different columns represent: the state, its latitude, longitude, and its name
# The first state is Alaska and the last state is Wyoming
# There are 52 total entries
print(df[df["state"] == "DC"])
print(df[df["state"] == "PR"])
# There are more than 50 states listed because "state_data.csv" includes information
# on the District of Columbia and Puerto Rico
```

```python
latitude = df["latitude"]
longitude = df["longitude"]
median_latitude = latitude.median()
median_longitude = longitude.median()
print(f"The median latitude is {median_latitude}, and the median longitude is {median_longitude}.")
```

```python
dist_to_med_lat = np.abs(latitude - 39.4355155)
dist_to_med_long = np.abs(longitude - (-89.093198))
df["distance to median latitude"] = dist_to_med_lat
df["distance to median longitude"] = dist_to_med_long
```

```python
df.shape
# There are still 52 rows, but now 6 columns!
```

```python
closest_to_med_lat = df["distance to median latitude"].min()
# print(closest_to_med_lat)
df_least_dtmlat = df[df["distance to median latitude"] == 0.11453550000000234]
# display(df_least_dtmlat)
# df_least_dtmlat.loc[5,"name"]
# df_least_dtmlat.loc[45,"name"]
print(f"The states closest to the median latitude are {df_least_dtmlat.loc[5,"name"]} and {df_least_dtmlat.loc[45,"name"]}. They are {np.round(closest_to_med_lat, 2)} away from the median latitude.")
```

```python
farthest_from_med_lat = df["distance to median latitude"].max()
df_greatest_dtmlat = df[df["distance to median latitude"] == farthest_from_med_lat]
# display(df_greatest_dtmlat)
print(f"The state farthest from the median latitude is {df_greatest_dtmlat.loc[0,"name"]}. It is {np.round(farthest_from_med_lat,2)} away from the median latitude.")
```

```python
closest_to_med_long = df["distance to median longitude"].min()
df_least_dtmlong = df[df["distance to median longitude"] == closest_to_med_long]
# display(df_least_dtmlong)
print(f"The states closest to the median longitude are {df_least_dtmlong.loc[14,"name"]}, {df_least_dtmlong.loc[25,"name"]}, and {df_least_dtmlong.loc[49,"name"]}. They are {np.round(closest_to_med_long,2)} away from the median longitude.")
```

```python
farthest_from_med_long = df['distance to median longitude'].max()
df_greatest_dtmlong = df[df['distance to median longitude'] == farthest_from_med_long]
# display(df_greatest_dtmlong)
print(f"The state farthest from the median longitude is {df_greatest_dtmlong.loc[11, 'name']}. It is {np.round(farthest_from_med_long,2)} away from the median longitude.")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
originaldata = sns.load_dataset("iris")
originaldata.head()
```

```python
data = originaldata.dropna()
data.shape
display(data.iloc[55:99])
```

```python
print(data.shape)
# The shape of the dataset is 150 x 5, meaning it has 150 rows and 5 columns

display(data.head())
display(data.tail())
      
print(data.columns)
# The column names are sepal length, sepal width, petal length, petal width, and species.
# These describe the qualities of various iris plant samples
```

```python
fig, axs = plt.subplots(2,2, figsize= (15,15))

setosa = data[data["species"] == "setosa"]
# display(setosa)
versicolor = data[data["species"] == "versicolor"]
# display(versicolor)
virginica = data[data["species"] == "virginica"]
# display(virginica)

set_sepal_length = setosa["sepal_length"]
set_sepal_width = setosa["sepal_width"]
ver_sepal_length = versicolor["sepal_length"]
ver_sepal_width = versicolor["sepal_width"]
vir_sepal_length = virginica["sepal_length"]
vir_sepal_width = virginica["sepal_width"]

axs[0,0].scatter(set_sepal_length, set_sepal_width, label = "setosa")
axs[0,0].scatter(ver_sepal_length, ver_sepal_width, label = "versicolor")
axs[0,0].scatter(vir_sepal_length, vir_sepal_width, label = "virginica")
axs[0,0].set_xlabel("Sepal Length")
axs[0,0].set_ylabel("Sepal Width")
axs[0,0].set_title("Sepal Length vs Width")
axs[0,0].legend()
# This top left subplot describes the sepal length and width of Iris Flowers, based on its species
# Plotting each species a different color allowed me to observe trends in each species' size
# For example, the Setosa species was in general shorter in length, but wider than 
# the other two species.

petal_length = data["petal_length"]
petal_width = data['petal_width']

axs[0,1].scatter(petal_length, petal_width, label = "data", color="gold")
axs[0,1].set_xlabel("Petal Length")
axs[0,1].set_ylabel("Petal Width")
axs[0,1].set_title("Petal Length vs Width")
axs[0,1].legend()
# The top right plot describes all Iris' general ratio of petal length to width
# Overall, for all species, petal width increases linearly with petal length.

sns.histplot(data=data, x='petal_length', ax=axs[1,0], color = "lavender")
axs[1,0].set_title('Histogram of Petal Length') 
# The lower left plot describes the number of each flower sample that had a given 
# petal length. 
# From this plot, I can easily conclude that the highest number of the Iris samples
# had petal length between 1 and 1.5 (units unclear, not included in given data).

sns.boxplot(data=data, x='petal_width', y= 'species', ax=axs[1,1], color = "red")
axs[1,1].set_title('Box Plot of Petal Width')
# The lower right plot is a box plot petal width and species. 
# For each Iris species, it describes the min, 1st quartile (where 25% of the lower data sits), 
# median, third quartile (where 25% of the higher data sits), and maximum of petal width
# From these plots I can conclude that the Setosa flowers had a smallest range of 
# petal width (range of 0.125: 0.25 to 0.375) compared to other species (where petal width ranged 0.5 units,
# from 1.75 to 2.25).

plt.show()
```
