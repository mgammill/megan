```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sin_wave(x, A, w):
    y = A * np.sin(w * x)
    plt.plot(x, y)
    plt.show()

x = np.linspace(-2 * np.pi, 2 * np.pi, 500)
make_sin_wave(x, 1, 1)
make_sin_wave(x, 4, 2)
```
