```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../python_decal_fa25/course_assignments/homework7/star_data.txt", skiprows=1)
data
```

```python
print(data.shape)
# 2642 rows, and 2 columns
# column 1: magnitude (brightness) of the star in units of magnitude
# column 2: temperature of the star in units Kelvin
```

```python
magnitude = data[:,0]
temperature = data[:,1]
print(f"Magnitude: {magnitude}")
print(f"Temperature: {temperature}")
print()

# Max and Min Magnitude
print(f"Maximum magnitude: {np.max(magnitude)}")
print(f"Maximum magnitude achieved in row: {np.where(magnitude == 12.49)[0]}")
print(f"This is the row: {data[264]}")
print("This is the dimmest star because it has the greatest magnitude")
print()

print(f"Minimum magnitude {np.min(magnitude)}")
print(f"Minimum magnitude achieved in row: {np.where(magnitude == 2.24)[0]}")
print(f"This is the row: {data[1988]}")
print("This is the brightest star becuase it has the lowest magnitude")
print()

# Max and Min Temperature
print(f"Maximum temperature: {np.max(temperature)}")
print(f"Maximum temperature achieved in row: {np.where(temperature == 10112.78057)[0]}")
print(f"This is the row: {data[2613]}")
print()

print(f"Minimum temperature: {np.min(temperature)}")
print(f"Minimum temperature achieved in row: {np.where(temperature == 3653.20966)[0]}")
print(f"This is the row: {data[1923]}")
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize = (8,6))
plt.scatter(log_temperature, log_magnitude, label = "Star")
dimmest_star = data[264]
plt.plot(np.log(dimmest_star[1]), np.log(dimmest_star[0]), markersize = 10, color = "purple", marker = "*", label = "Dimmest Star")
brightest_star = data[1988]
plt.plot(np.log(brightest_star[1]), np.log(brightest_star[0]), markersize= 10, color = "black", marker = "*", label = "Brightest Star")
hottest_star = data[2613]
plt.plot(np.log(hottest_star[1]), np.log(hottest_star[0]), markersize=10, color = "green", marker = "*", label = "Hottest Star")
coldest_star = data[1923]
plt.plot(np.log(coldest_star[1]), np.log(coldest_star[0]), markersize =10, color = "orange", marker = "*", label = "Coldest Star")
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.title("Hertzspring-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.legend()
plt.show()
```
