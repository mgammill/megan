```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sin_wave(x, A, w):
    y = A * np.sin(w * x)
    plt.plot(x, y)
    plt.show()

x = np.linspace(-2 * np.pi, 2 * np.pi, 500)
make_sin_wave(x, 1, 1)
make_sin_wave(x, 4, 2)
```

```python
x = np.linspace(0, 2 * np.pi, 1000)
amplitude = np.array([0.5, 1, 1.5, 2, 2.5])
frequency = np.array([1, 2, 3, 4, 5])
```

```python
fig, ax = plt.subplots(1, 3, figsize = (25, 5))
plt.show()
```

```python
for x, y in zip(amplitude, frequency):
    print(x, y)
```

```python
fig, ax = plt.subplots(1, 3, figsize = (25, 5))
x = np.linspace(0,2 * np.pi, 1000)
for A in zip(amplitude):
    ax[0].plot(x, A * np.sin(x), label = f"Amplitude {float(A[0])}")
    ax[0].set_title("Varying Amplitude", fontsize=20)
    ax[0].set_xlabel("x", fontsize=18)
    ax[0].set_ylabel("Amplitude * sin(x)", fontsize=18)
    ax[0].legend(loc = "best", fontsize=12)
for w in zip(frequency):
    ax[1].plot(x, np.sin(w * x), label = f"Wavelength {float(w[0])}")
    ax[1].set_title("Varying Wavelength", fontsize=20)
    ax[1].set_xlabel("x", fontsize=18)
    ax[1].set_ylabel("sin(Wavelength * x)", fontsize=18)
    ax[1].legend(loc = "best", fontsize=12)
for A, w in zip(amplitude, frequency):
    ax[2].plot(x, A * np.sin(w * x), label = f"Amplitude {A}, Wavelength {w}")
    ax[2].set_title("Varying Amplitude and Wavelength", fontsize=20)
    ax[2].set_xlabel("x", fontsize=18)
    ax[2].set_ylabel("Amplitude * sin(Wavelength * x)", fontsize=18)
    ax[2].legend(loc = "best", fontsize=12)
plt.show()
```
